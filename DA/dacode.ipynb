```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img,img_to_array

datagen=ImageDataGenerator(
    rotation_range=40,
    height_shift_range=0.4,
    width_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=0.4,
    rescale=None,
    fill_mode='reflect'
    

)
img=load_img(r"C:\Users\vamsikrishna\OneDrive\Pictures\WhatsApp Image 2026-10-29 at 10.08.52 PM.jpeg")
img
```

```python
x = img_to_array(img)
x = x.reshape((1,) + x.shape)  # Reshape to (1, height, width, channels)

# Generate augmented images and save
i = 0
for batch in datagen.flow(x, batch_size=1,
                          save_to_dir=r"C:\Users\vamsikrishna\OneDrive\Documents\projects\DA",
                          save_prefix='krishna', save_format='jpeg'):
    i += 1
    if i > 30:
        break  
```
